# Problema 1: Clasificador de temperatura

```python


# range(-30, 51) genera números desde -30 hasta 50 inclusive.
for numero in range(-30, 51):
    if numero < 0:
        print(f"{numero} Muy frío")
    elif 0 <= numero <= 24:
        print(f"{numero} Templado")
    else:
        print(f"{numero} Caluroso")
```

# Problema 2: Números especiales del robot

```python


# range(1, 101) va desde el 1 hasta el 100.
for numero in range(1, 101):
    # % es el operador de resto. Si numero % 5 == 0 significa "es divisible por 5".
    if numero % 5 == 0:
        if numero % 10 == 0:
            print(f"{numero} Bip Bip")
        else:
            print(f"{numero} Bip")
```

# Problema 3: Detector de puntaje gamer

```python


# input() le pide al usuario que ingrese un dato.
# int() convierte el texto ingresado en un número para poder sumarlo.
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

suma = puntaje1 + puntaje2

if suma < 50:
    resultado = "Nivel principiante"
elif 50 <= suma < 80:
    resultado = "Nivel intermedio"
else:
    resultado = "Nivel pro"

print(f"\nResultado: {resultado}")
```
