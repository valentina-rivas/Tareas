# Problema 1: Clasificador de temperatura

```python

# Problema 1: Clasificador de temperatura

# Usamos un bucle for para recorrer los números uno a uno.
# range(-30, 51) genera la secuencia desde -30 hasta 50 inclusive.
for numero in range(-30, 51):

    # Caso 1: Si el número es estrictamente menor a 0
    if numero < 0:
        print(f"{numero} Muy frío")

    # Caso 2: Si el número está en el rango de 0 a 24 inclusive
    elif 0 <= numero <= 24:
        print(f"{numero} Templado")

    # Caso 3: Si el número es igual o mayor a 25
    else:
        print(f"{numero} Caluroso")
```

# Problema 2: Números especiales del robot

```python


# Recorremos los números del 1 al 100 inclusive.
for numero in range(1, 101):

    # Filtramos únicamente los números que son divisibles por 5.
    # El operador % calcula el resto de la división (si es 0, es divisible).
    if numero % 5 == 0:

        # Condición especial: Si además de ser divisible por 5, también es divisible por 10
        if numero % 10 == 0:
            print(f"{numero} Bip Bip")

        # Si solo es divisible por 5 pero no por 10
        else:
            print(f"{numero} Bip")
```

# Problema 3: Detector de puntaje gamer

```python


# Solicitamos los datos al usuario.
# input() recibe el texto ingresado y int() lo transforma a número entero para poder operarlo.
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Sumamos los dos puntajes obtenidos
suma_total = puntaje1 + puntaje2

# Evaluamos la suma total según los rangos establecidos:

# Caso 1: Si la suma es estrictamente menor que 50
if suma_total < 50:
    resultado = "Nivel principiante"

# Caso 2: Si la suma está entre 50 inclusive y menor que 80
elif 50 <= suma_total < 80:
    resultado = "Nivel intermedio"

# Caso 3: Si la suma es mayor o igual a 80
else:
    resultado = "Nivel pro"

# Mostramos el resultado por pantalla
print(f"\nResultado: {resultado}")
```

# Problema 4: Calculador de perfil musical

```python


# Creamos una función que recibe los tres parámetros requeridos y retorna el mensaje correspondiente.
# edad: número entero
# le_gusta_musica y le_gusta_bailar: valores booleanos (True o False)
def calculador_perfil_musical(edad, le_gusta_musica, le_gusta_bailar):

    # Caso 1: Tener 18 años o más, gustarle la música y gustarle bailar
    if edad >= 18 and le_gusta_musica and le_gusta_bailar:
        return "Eres una estrella de la fiesta"

    # Caso 2: Le gusta la música pero NO le gusta bailar
    elif le_gusta_musica and not le_gusta_bailar:
        return "Te gusta la música, pero no tanto bailar"

    # Caso 3: Cualquier otra combinación posible
    else:
        return "Tienes tu propio estilo"


# Le pedimos la edad al usuario
edad = int(input("¿Cuántos años tienes? "))

# Solicitamos las respuestas de sí/no y las preprocesamos
# .strip() elimina espacios extra y .lower() pasa el texto a minúsculas
respuesta_musica = input("¿Te gusta la música? (si/no): ").strip().lower()
respuesta_bailar = input("¿Te gusta bailar? (si/no): ").strip().lower()

# Convertimos la respuesta de texto en valor booleano True o False
le_gusta_musica = respuesta_musica == "si" or respuesta_musica == "sí"
le_gusta_bailar = respuesta_bailar == "si" or respuesta_bailar == "sí"

# Ejecutamos la función y mostramos la evaluación final en pantalla
print(calculador_perfil_musical(edad, le_gusta_musica, le_gusta_bailar))
```
